```python
import os
from astropy.io import fits

short_path = "/Volumes/morgana2/snert/max/data/JWST/"
path = f"{short_path}downloads/JWST"

uncal_dict = {}

for dir in os.listdir(path):
    if not dir.startswith("jw"):
        continue
    filename = os.listdir(os.path.join(path, dir))[0]
    if filename.startswith("."):
        continue
    full_path = os.path.join(path, dir, filename)

    hdu = fits.open(full_path)

    prog_id = hdu[0].header["PROGRAM"][1:]
    print(prog_id)

    # if prog_id == "1242":
    #     continue

    if prog_id not in uncal_dict.keys():
        uncal_dict[prog_id] = []
    uncal_dict[prog_id].append((hdu, full_path, filename))
```
